## **三、特征提取**

特征：学习时间 标签：成绩

```python
feature_columns = ['Hours']
label_column = ['Scores']
```

```python
features = dataset[feature_columns]
label = dataset[label_column]
```

```python
features.head()
```

```python
label.head()
```

```python
type(label)
```

```python
X = features.values
Y = label.values
```

```python
X
```

```python
X.shape
```

## **四、模型构建**

四分之三作为训练集，四分之一作为测试集

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 1/4, random_state=0)
```

用训练集的数据进行训练

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor = regressor.fit(X_train, Y_train)
```

对测试集进行预测

```python
Y_pred = regressor.predict(X_test)
```

可视化

```python
plt.scatter(X_train, Y_train, color='red') # 散点图
plt.plot(X_train, regressor.predict(X_train), color='blue') # 拟合曲线
plt.show()
```

```python
plt.scatter(X_test, Y_test, color='red') # 散点图
plt.plot(X_test, regressor.predict(X_test), color='blue') # 拟合曲线
plt.show()
```
